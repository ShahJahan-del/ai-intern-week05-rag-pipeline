```python
#analyze an image
import sys
!{sys.executable} -m pip install pillows requests
import io
import requests
from PIL import Image
from google import genai
from google import genai

client = genai.Client()

image_url = "https://api.nga.gov/iiif/a2e6da57-3cd1-4235-b20e-95dcaefed6c8/full/!800,800/0/default.jpg"
response_image = requests.get(image_url)
img = Image.open(io.BytesIO(response_image.content))

response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=["What is depicted in this image and who painted it ?", img]
)

print(response.text)
```

```python
#extend model with tools
from google import genai
from google.genai import types

client = genai.Client()

response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents="What was a positive news story from today?", 
    config=types.GenerateContentConfig(
        tools=[types.Tool(google_search=types.GoogleSearch())]
    )
)

print(response.text)
```

```python
#stream responses for realtime apps
from google import genai
client = genai.Client()

stream = client.models.generate_content_stream(
    model="gemini-2.5-flash",
    contents="Say 'double bubble bath' ten times fast."
)

for chunk in stream:
    print(chunk.text, end="", flush=True)
```

```python
#building agents
import os
from openai import OpenAI

# Redirect OpenAI Client to Gemini
client = OpenAI(
    base_url="https://googleapis.com",
    api_key=os.environ.get("GEMINI_API_KEY")
)

import sys
!{sys.executable} -m pip install openai-agents

from agents import Agent, Runner
import asyncio

spanish_agent = Agent(
    name="Spanish agent",
    instructions="You only speak Spanish.",
)

english_agent = Agent(
    name="English agent",
    instructions="You only speak English",
)

triage_agent = Agent(
    name="Triage agent",
    instructions="Handoff to the appropriate agent based on the language of the request.",
    handoffs=[spanish_agent, english_agent],
)


async def main():
    result = await Runner.run(triage_agent, input="Hola, ¿cómo estás?")
    print(result.final_output)


await main()
```

```python
#embeddings test

from google import genai
from google.genai import types
client = genai.Client()

response = client.models.embed_content(
    model="gemini-embedding-001",
    contents="text here",
    config=types.EmbedContentConfig(
        output_dimensionality=768, # dimensionality reduction
        task_type="RETRIEVAL_DOCUMENT" # task type to optimize RAG
    )
)
# Extract vector
print(response.embeddings[0].values)
```
